````python
def understand_question(question):
    question = question.lower().strip()

    if "sub-categor" in question and "profit" in question:
        return "top_subcategories_by_profit"

    elif "category" in question and "profit" in question:
        return "profit_by_category"

    elif "region" in question and "sales" in question:
        return "sales_by_region"

    else:
        return "unknown"


def analyze_data(df, analysis_type):

    if analysis_type == "profit_by_category":
        return (
            df.groupby("Category")["Profit"]
              .sum()
              .sort_values(ascending=False)
        )

    elif analysis_type == "sales_by_region":
        return (
            df.groupby("Region")["Sales"]
              .sum()
              .sort_values(ascending=False)
        )

    elif analysis_type == "top_subcategories_by_profit":
        return (
            df.groupby("Sub-Category")["Profit"]
              .sum()
              .sort_values(ascending=False)
              .head(10)
        )

    else:
        return None


def generate_answer(result, analysis_type):

    if result is None:
        return "I don't know how to answer this question yet."

    if analysis_type == "profit_by_category":
        return (
            f"{result.index[0]} has the highest total profit "
            f"at {result.iloc[0]:,.2f}."
        )

    elif analysis_type == "sales_by_region":
        return (
            f"{result.index[0]} has the highest total sales "
            f"at {result.iloc[0]:,.2f}."
        )

    elif analysis_type == "top_subcategories_by_profit":
        return "Top 10 sub-categories by profit:\n\n" + result.to_string()


def ask_agent(df, question):

    analysis_type = understand_question(question)
    result = analyze_data(df, analysis_type)
    return generate_answer(result, analysis_type)
````

````python
# ============================================================
# FINAL AI DATA ANALYST AGENT
# ============================================================

import json
import re
import pandas as pd


# ------------------------------------------------------------
# 1. AI PLANNER
# ------------------------------------------------------------

def ask_ai(question):

    prompt = f"""
You are a data analysis planning assistant.

Dataset columns:
{list(df.columns)}

User question:
{question}

Create a NEW analysis plan specifically for this question.

Return ONLY valid JSON.
Do not return markdown.
Do not return explanations.

Rules:

- category -> group_by = "Category"
- region -> group_by = "Region"
- sub-category -> group_by = "Sub-Category"

- profit -> metric = "Profit"
- sales -> metric = "Sales"
- quantity -> metric = "Quantity"

- highest / largest / maximum -> sort = "descending"
- lowest / smallest / minimum -> sort = "ascending"

- total -> operation = "sum"
- average / avg -> operation = "mean"
- how many / count / number of records -> operation = "count"

For count:
DO NOT include metric.

For top N:
limit = N

For bottom N:
limit = N and sort = "ascending"

For conditions such as:
"in the West region"
"in the East region"
"for Technology"
"for Furniture"

use:

"filter": {{
    "column": "...",
    "value": "..."
}}

Examples:

Question:
Which category has the highest profit?

JSON:
{{
    "group_by": "Category",
    "metric": "Profit",
    "operation": "sum",
    "sort": "descending",
    "limit": 1
}}

Question:
Which region has the highest sales?

JSON:
{{
    "group_by": "Region",
    "metric": "Sales",
    "operation": "sum",
    "sort": "descending",
    "limit": 1
}}

Question:
Which sub-category has the lowest profit?

JSON:
{{
    "group_by": "Sub-Category",
    "metric": "Profit",
    "operation": "sum",
    "sort": "ascending",
    "limit": 1
}}

Question:
How many records are in each region?

JSON:
{{
    "group_by": "Region",
    "operation": "count"
}}

Question:
What are the top 3 sub-categories by profit?

JSON:
{{
    "group_by": "Sub-Category",
    "metric": "Profit",
    "operation": "sum",
    "sort": "descending",
    "limit": 3
}}

Question:
Which category has the highest profit in the West region?

JSON:
{{
    "filter": {{
        "column": "Region",
        "value": "West"
    }},
    "group_by": "Category",
    "metric": "Profit",
    "operation": "sum",
    "sort": "descending",
    "limit": 1
}}

Return ONLY JSON.
"""

    response = ollama.chat(
        model="qwen2.5:3b",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return response["message"]["content"]


# ------------------------------------------------------------
# 2. PARSE AI PLAN
# ------------------------------------------------------------

def parse_plan(ai_response):

    text = ai_response.strip()

    # Remove markdown fences if the model adds them
    text = text.replace("```json", "").replace("```", "").strip()

    # Extract JSON object if extra text appears
    match = re.search(r"\{.*\}", text, re.DOTALL)

    if not match:
        raise ValueError("AI did not return valid JSON.")

    return json.loads(match.group())


# ------------------------------------------------------------
# 3. EXECUTE PLAN
# ------------------------------------------------------------

def execute_plan(data, plan):

    working_df = data.copy()

    group_by = plan.get("group_by")
    metric = plan.get("metric")
    operation = str(plan.get("operation", "")).lower()
    sort_order = plan.get("sort", "descending")
    limit = plan.get("limit")
    filter_info = plan.get("filter")

    # Apply filter
    if filter_info:

        filter_column = filter_info.get("column")
        filter_value = filter_info.get("value")

        if filter_column not in working_df.columns:
            raise ValueError(
                f"Invalid filter column: {filter_column}"
            )

        working_df = working_df[
            working_df[filter_column] == filter_value
        ]

    # Validate group
    if group_by not in working_df.columns:
        raise ValueError(
            f"Invalid group-by column: {group_by}"
        )

    # Count
    if operation == "count":

        result = working_df.groupby(group_by).size()

    else:

        if metric not in working_df.columns:
            raise ValueError(
                f"Invalid metric column: {metric}"
            )

        if operation in ["avg", "average"]:
            operation = "mean"

        if operation == "sum":

            result = (
                working_df
                .groupby(group_by)[metric]
                .sum()
            )

        elif operation == "mean":

            result = (
                working_df
                .groupby(group_by)[metric]
                .mean()
            )

        else:
            raise ValueError(
                f"Unsupported operation: {operation}"
            )

    # Sort
    if sort_order == "ascending":
        result = result.sort_values(ascending=True)
    else:
        result = result.sort_values(ascending=False)

    # Limit
    if limit is not None:
        result = result.head(int(limit))

    return result


# ------------------------------------------------------------
# 4. VERIFY RESULT
# ------------------------------------------------------------

def verify_result(question, result, plan, data):

    try:

        expected = execute_plan(data, plan)

        # Compare indexes and values
        if isinstance(expected, pd.Series) and isinstance(result, pd.Series):

            expected = expected.sort_index()
            actual = result.sort_index()

            values_match = expected.index.equals(actual.index)

            if values_match:
                values_match = all(
                    abs(float(a) - float(b)) < 0.000001
                    for a, b in zip(expected.values, actual.values)
                )

            verified = values_match

        else:
            verified = expected.equals(result)

        return {
            "verified": bool(verified),
            "expected": expected,
            "actual": result
        }

    except Exception as e:

        return {
            "verified": False,
            "error": str(e)
        }


# ------------------------------------------------------------
# 5. BUSINESS INSIGHT
# ------------------------------------------------------------

def generate_business_insight(question, result, verification):

    if verification.get("verified") is not True:
        return "Business insight unavailable because the result could not be verified."

    q = question.lower()

    # Single result
    if len(result) == 1:

        name = result.index[0]
        value = result.iloc[0]

        if "profit" in q:
            metric_name = "profit"
        elif "sales" in q:
            metric_name = "sales"
        elif "quantity" in q:
            metric_name = "quantity"
        elif "record" in q or "count" in q:
            metric_name = "records"
        else:
            metric_name = "value"

        if metric_name == "records":
            return (
                f"{name} has {int(value):,} records."
            )

        return (
            f"{name} has the highest total {metric_name}, "
            f"with {float(value):,.2f}."
            if any(word in q for word in ["highest", "top", "largest"])
            else
            f"{name} has the lowest total {metric_name}, "
            f"with {float(value):,.2f}."
        )

    # Multiple results
    if len(result) > 1:

        if "profit" in q:
            metric_name = "profit"
        elif "sales" in q:
            metric_name = "sales"
        elif "quantity" in q:
            metric_name = "quantity"
        elif "record" in q or "count" in q:
            metric_name = "records"
        else:
            metric_name = "value"

        items = [
            f"{name} ({float(value):,.2f})"
            for name, value in result.items()
        ]

        return (
            f"The results by {result.index.name} are: "
            + ", ".join(items)
            + "."
        )

    return "Verified result available."


# ------------------------------------------------------------
# 6. MAIN AGENT
# ------------------------------------------------------------

def final_ai_data_analyst(question):

    print("\n" + "=" * 70)
    print("AI DATA ANALYST AGENT")
    print("=" * 70)

    print("\nQUESTION:")
    print(question)

    try:

        # AI creates plan
        ai_response = ask_ai(question)

        # Convert AI response to JSON
        plan = parse_plan(ai_response)

        print("\nAI PLAN:")
        print(json.dumps(plan, indent=4))

        # Execute analysis
        result = execute_plan(df, plan)

        print("\nRESULT:")
        print(result)

        # Verify against Pandas
        verification = verify_result(
            question,
            result,
            plan,
            df
        )

        print("\nVERIFICATION:")

        if verification["verified"]:
            print("✅ PASSED")
        else:
            print("❌ FAILED")

        # Business insight
        insight = generate_business_insight(
            question,
            result,
            verification
        )

        print("\nBUSINESS INSIGHT:")
        print(insight)

        return {
            "question": question,
            "plan": plan,
            "result": result,
            "verification": verification,
            "business_insight": insight
        }

    except Exception as e:

        print("\n❌ ERROR:")
        print(e)

        return {
            "question": question,
            "error": str(e)
        }


print("✅ FINAL AI DATA ANALYST AGENT LOADED")
````

````python
# AI Data Analyst Agent

An AI-powered data analysis agent that converts natural-language business questions into structured analytical plans, executes them using Pandas, verifies the results programmatically, and generates concise business insights.

## Project Workflow

User Question  
↓  
AI Planner  
↓  
JSON Analysis Plan  
↓  
Pandas Execution  
↓  
Result Verification  
↓  
Business Insight

## Key Capabilities

- Natural-language data analysis
- JSON-based analysis planning
- Grouping and aggregation
- Sum, average, and count operations
- Sorting and Top-N analysis
- Conditional filtering
- Programmatic result verification
- Automated business insights
- Error handling

## Dataset

Retail sales dataset containing 9,994 records and 13 columns.

## Technologies

- Python
- Pandas
- Ollama
- Qwen 2.5 3B
- Jupyter Notebook
- JSON

## Example Questions

- Which category has the highest profit?
- Which region has the highest sales?
- Which sub-category has the lowest profit?
- How many records are in each region?
- What are the top 3 sub-categories by profit?
- Which category has the highest profit in the West region?
````

````python
## Architecture

```text
                    ┌─────────────────────┐
                    │    User Question    │
                    └──────────┬──────────┘
                               │
                               ▼
                    ┌─────────────────────┐
                    │     AI Planner      │
                    │   Qwen 2.5 3B       │
                    └──────────┬──────────┘
                               │
                               ▼
                    ┌─────────────────────┐
                    │    JSON Plan        │
                    │ group_by / metric   │
                    │ operation / filter  │
                    │ sort / limit        │
                    └──────────┬──────────┘
                               │
                               ▼
                    ┌─────────────────────┐
                    │   Pandas Engine     │
                    │   Data Processing   │
                    └──────────┬──────────┘
                               │
                               ▼
                    ┌─────────────────────┐
                    │ Result Verification │
                    │  Independent Check  │
                    └──────────┬──────────┘
                               │
                               ▼
                    ┌─────────────────────┐
                    │  Business Insight   │
                    └─────────────────────┘
````

````python
## Limitations

- The current agent is designed for structured analytical questions that map to supported Pandas operations.
- Supported operations include sum, average, and count.
- Filtering currently focuses on categorical conditions such as region and category.
- The agent operates on the provided retail dataset and does not connect to external databases or live data sources.
- Business insights are generated only after the analytical result passes programmatic verification.
- Complex natural-language questions outside the supported analytical patterns may require additional planning rules.
````
